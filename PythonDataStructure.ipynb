# set ;- Unordered,mutable and no duplicates

```python
myset = {10,20,30}  # Add new data without function
myset = myset | {40};
print(myset);
```

```python
myset = {10,20,30,40,50,20} 
print(myset);
print(type(myset));
```

```python
myset = {10,20,30,40} # Remove data without function
myset = myset - {10}
print(myset);
```

```python
set1 = {1,2,3}
set2 = {4,5,6}

newset = set1 | set2;
print(newset);
```

```python
set1 = {1,2,3,4}
set2 = {3,4,5,6}

commondata = set1 & set2
print(commondata)
```

```python
myset = {10,20,20,30,30,40}
print(myset)
```

```python
myset = {10,20,30}
myset.add(40)
print(myset);
```

```python
myset = {10,20,30,40}
myset.remove(20)
print(myset);
```

```python
set1 = {1,2,3}
set2 = {3,4,5}
print(set1.union(set2));
```

```python
set1 = {1,2,3,4};
set2 = {3,4,5,6};
print(set1.intersection(set2));
```

```python
set1 = {1,2,3};
set2 = {3,4,5};
print(set1.symmetric_difference(set2));
```
